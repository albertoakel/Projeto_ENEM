# 06 — Análise Transversal

## Objetivo

Este notebook realiza uma análise exploratória transversal (fotografia de um único ano) dos dados da camada `Consumo`. O foco é comparar os municípios da UF selecionada em relação à participação no ENEM, ao desempenho nas disciplinas disponíveis e ao perfil dos participantes.

A análise está organizada em três eixos: panorama municipal; desempenho por grupos (dependência administrativa da escola, sexo e faixa etária); e síntese dos resultados do ano.

### Regra metodológica central: compatibilidade temporal

| Período | Disciplinas analisadas |
|---|---|
| 1998–2008 | `OBJETIVA` e `REDACAO_ANTIGA` |
| 2009–2025 | `CN`, `CH`, `LC`, `MT` e `REDACAO_NOVA` |

As versões antiga e nova da Redação não são tratadas como uma série homogênea. `configuracao_analise_transversal()` define as disciplinas, os nomes e as colunas compatíveis com o ano escolhido. As análises por escola, sexo e faixa etária seguem a mesma seleção temporal.

### Como utilizar

- Ajuste `ANO_TRANSVERSAL` e `UF_PROCESSAMENTO` na seção 1.
- Execute as células em sequência: as seções posteriores reutilizam objetos calculados anteriormente.
- Observe os filtros mínimos descritos em cada seção.
- Os resultados são descritivos e exploratórios; diferenças observadas não demonstram, isoladamente, relações causais.


## 1. Configuração e carga

Define o recorte geográfico e temporal e carrega os produtos da camada `Consumo`.

- `ANO_TRANSVERSAL`: ano que será analisado; a análise representa uma fotografia desse ano.
- `UF_PROCESSAMENTO`: unidade da Federação selecionada.
- `configuracao_analise_transversal()`: determina o período metodológico, as disciplinas válidas, as colunas de médias e seus rótulos.
- `read_allfiles2()`: carrega os produtos municipais utilizados no notebook.

| Objeto | Produto | Uso |
|---|---|---|
| `painel` | `painel_municipal` | Participação, médias e indicadores municipais integrados |
| `faixa` | `faixa_etaria_municipal` | Participantes por faixa etária e município |
| `grupo` | `desempenho_grupos_municipal` | Desempenho por escola, sexo e faixa etária |

As dimensões impressas ao final permitem conferir a carga das três bases.


```python
# ============================================================
# 1. CONFIGURAÇÃO E CARGA
# ============================================================
from _setup import *
import unicodedata
import warnings
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from enem_pipeline.config import (
    PROJECT_ROOT, DATA_DIR, METADATA_DIR, CONSUMO_DIR,
    UF_PROCESSAMENTO, ANO_INICIAL, ANO_FINAL
)
from utils.features_names import read_allfiles,read_allfiles2
from utils.eda import *
import time

warnings.filterwarnings("ignore", category=FutureWarning, module="seaborn")
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 160)

# ============================================================
# PARÂMETROS DA ANÁLISE TRANSVERSAL
# ============================================================
ANO_TRANSVERSAL = 2025
UF_PROCESSAMENTO = "PA"

config_ano = configuracao_analise_transversal(ANO_TRANSVERSAL)
PERIODO_TRANSVERSAL = config_ano["periodo"]
PERIODO_ID_TRANSVERSAL = config_ano["periodo_id"]
DISCIPLINAS_TRANSVERSAL = config_ano["disciplinas"]
colunas_medias = config_ano["colunas_medias"]
nomes_disciplinas = config_ano["nomes_disciplinas"]
ORDEM_DISCIPLINAS = config_ano["ordem_disciplinas"]
colunas_top5 = nomes_disciplinas

out = read_allfiles2(UF_PROCESSAMENTO,'ESC')
painel = out["painel"]["data"]
faixa = out["faixa"]["data"]
grupo = out["grupo"]["data"]

print("=" * 80)
print("ANÁLISE TRANSVERSAL")
print("=" * 80)
print(f"Ano transversal: {ANO_TRANSVERSAL}")
print(f"UF: {UF_PROCESSAMENTO}")
print(f"Período metodológico: {PERIODO_TRANSVERSAL['rotulo']}")
print(f"Estrutura de desempenho: {PERIODO_TRANSVERSAL['estrutura']}")
print(f"Disciplinas incluídas: {', '.join(ORDEM_DISCIPLINAS)}")
print(f"Painel: {painel.shape[0]:,} linhas × {painel.shape[1]} colunas")
print(f"Faixa:  {faixa.shape[0]:,} linhas × {faixa.shape[1]} colunas")
print(f"Grupo:  {grupo.shape[0]:,} linhas × {grupo.shape[1]} colunas")

```

## 2. Disponibilidade do ano transversal

Verifica separadamente se o ano selecionado está presente em `painel`, `faixa` e `grupo`.

A ausência do ano em `painel_municipal` interrompe a análise, pois esse produto é a base principal. Se o ano não estiver em `desempenho_grupos_municipal`, o notebook continua, mas as análises por escola, sexo e faixa etária ficam indisponíveis. Não há substituição automática por outro ano.


```python
# ============================================================
# 2. DISPONIBILIDADE DO ANO TRANSVERSAL
# ============================================================
anos_painel = set(pd.to_numeric(painel["ano"], errors="coerce").dropna().astype(int))
anos_faixa = set(pd.to_numeric(faixa["ano"], errors="coerce").dropna().astype(int))
anos_grupo = set(pd.to_numeric(grupo["ano"], errors="coerce").dropna().astype(int))

print(f"Ano selecionado: {ANO_TRANSVERSAL} — período {PERIODO_TRANSVERSAL['rotulo']}")
print(f"painel_municipal: {ANO_TRANSVERSAL in anos_painel}")
print(f"faixa_etaria_municipal: {ANO_TRANSVERSAL in anos_faixa}")
print(f"desempenho_grupos_municipal: {ANO_TRANSVERSAL in anos_grupo}")

TEM_GRUPOS_ANO = ANO_TRANSVERSAL in anos_grupo
if ANO_TRANSVERSAL not in anos_painel:
    raise ValueError(f"O painel municipal não contém o ano {ANO_TRANSVERSAL}.")
if not TEM_GRUPOS_ANO:
    print(
        f"ATENÇÃO: desempenho_grupos_municipal não possui {ANO_TRANSVERSAL}. "
        "As análises por escola, sexo e faixa etária ficarão indisponíveis; "
        "nenhum outro ano será usado como substituto."
    )
```

## 3. Panorama do ano transversal

Seleciona no painel municipal apenas o ano definido e verifica se as colunas de desempenho esperadas para o período estão disponíveis.

Apresenta a quantidade de municípios representados e o total de participantes. A tabela `panorama` resume cada disciplina pela média e mediana das médias municipais. Portanto, essas estatísticas descrevem a distribuição entre municípios, não a média individual de todos os participantes do estado.


```python
# ============================================================
# 3. PANORAMA DO ANO TRANSVERSAL
# ============================================================
dados_ano = painel.loc[pd.to_numeric(painel["ano"], errors="coerce") == ANO_TRANSVERSAL].copy()
colunas_ausentes = [c for c in colunas_medias if c not in dados_ano.columns]
if colunas_ausentes:
    raise KeyError(
        f"Colunas esperadas para {PERIODO_TRANSVERSAL['rotulo']} não encontradas: "
        f"{colunas_ausentes}"
    )

print(f"Ano analisado: {ANO_TRANSVERSAL}")
print(f"Período metodológico: {PERIODO_TRANSVERSAL['rotulo']}")
print(f"Municípios com registros: {dados_ano['CO_MUNICIPIO_TERRITORIO'].nunique():,}")
print(f"Participantes: {dados_ano['n_participantes'].sum():,.0f}")

panorama = pd.DataFrame({
    "disciplina": [nomes_disciplinas[c] for c in colunas_medias],
    "media_municipal": [dados_ano[c].mean() for c in colunas_medias],
    "mediana_municipal": [dados_ano[c].median() for c in colunas_medias],
})
display(panorama.round(2))
```

## 4. Distribuição municipal do desempenho

Detalha a distribuição das médias municipais em cada disciplina válida. Para cada disciplina, calcula quantidade de municípios com dados, média, mediana, desvio-padrão, primeiro e terceiro quartis, mínimo e máximo.

A tabela permite avaliar a heterogeneidade territorial do desempenho. A unidade estatística é o município, e não o participante individual.


```python
# ============================================================
# 4. DISTRIBUIÇÃO MUNICIPAL DO DESEMPENHO
# ============================================================
linhas = []
for coluna, disciplina in nomes_disciplinas.items():
    valores = dados_ano[coluna].dropna()
    if valores.empty:
        continue
    linhas.append({
        "disciplina": disciplina, "n_municipios": len(valores),
        "media": valores.mean(), "mediana": valores.median(),
        "std": valores.std(), "q1": valores.quantile(.25),
        "q3": valores.quantile(.75), "min": valores.min(), "max": valores.max()
    })
distribuicao_desempenho = pd.DataFrame(linhas).sort_values("disciplina").reset_index(drop=True)
display(distribuicao_desempenho.round(2))
```

## 5. Top 5 — Número de Participantes

Identifica os cinco municípios com maior número de participantes no ano selecionado.

O ranking utiliza `n_participantes` do painel municipal e não aplica o limiar mínimo usado nas seções de desempenho. Representa concentração de participação, não classificação de desempenho.


```python
# ============================================================
# 5. TOP 5 — NÚMERO DE PARTICIPANTES
# ============================================================

top5_participantes = (dados_ano[["NO_MUNICIPIO_TERRITORIO","n_participantes"]].sort_values("n_participantes",ascending=False).head(5).reset_index(drop=True))

display(top5_participantes)
```

## 6. Top 5 — Desempenho por notas

Identifica os municípios com maiores resultados em cada disciplina válida, considerando separadamente média e mediana municipais.

**Elegibilidade:** o município precisa apresentar média, mediana e quantidade de notas válidas para a disciplina. São considerados apenas municípios com `n_validos > 100`; o filtro é aplicado antes da ordenação.

O código constrói `top5_media` e `top5_mediana`. A exibição atual apresenta o Top 5 por média; o ranking por mediana permanece calculado e disponível no objeto correspondente. A comparação entre média e mediana ajuda a observar a sensibilidade do ranking à distribuição dos resultados.


```python
## ============================================================
# 6. TOP 5 — DESEMPENHO POR MÉDIA E MEDIANA -- NOVO
# ============================================================
N_VALIDOS_MIN = 100  # ← limiar mínimo de participantes válidos

top5_media, top5_mediana = {}, {}

for coluna, disciplina in colunas_top5.items():
    sufixo = coluna.replace("media_", "")
    coluna_mediana  = coluna.replace("media_", "mediana_")
    coluna_validos  = f"n_validos_{sufixo}"

    necessarias = ["NO_MUNICIPIO_TERRITORIO", coluna_validos, coluna, coluna_mediana]
    faltantes = [c for c in necessarias if c not in dados_ano.columns]
    if faltantes:
        print(f"Indicador {disciplina} ignorado; colunas ausentes: {faltantes}")
        continue

    # ---- base comum: precisa ter média, mediana e N válidos > 100 ----
    base = dados_ano[necessarias].dropna(subset=[coluna, coluna_mediana])
    base = base[base[coluna_validos] > N_VALIDOS_MIN]

    if base.empty:
        print(f"Indicador {disciplina} sem municípios com N válidos > {N_VALIDOS_MIN}.")
        continue

    # ---- Top 5 por MÉDIA ----
    top5_media[disciplina] = (
        base.sort_values(coluna, ascending=False)
            .head(5)
            .reset_index(drop=True)
    )

    # ---- Top 5 por MEDIANA ----
    top5_mediana[disciplina] = (
        base.sort_values(coluna_mediana, ascending=False)
            .head(5)
            .reset_index(drop=True)
    )

# ---- exibição ----
for disciplina in top5_media:
    print(f"\nTOP 5 — {disciplina} — MÉDIA (N válidos > {N_VALIDOS_MIN})")
    tabela = top5_media[disciplina].copy()
    tabela.columns = ["Município", "N válidos", "Média", "Mediana"]
    display(tabela.round(2))

    # print(f"TOP 5 — {disciplina} — MEDIANA (N válidos > {N_VALIDOS_MIN})")
    # tabela = top5_mediana[disciplina].copy()
    # tabela.columns = ["Município", "N válidos", "Média", "Mediana"]
    # display(tabela.round(2))
```

### 6.1. Gráfico — Top 5 municípios por média das notas

Apresenta os municípios mais bem posicionados segundo a média municipal de cada disciplina, em painéis separados. A seleção de disciplinas respeita o período metodológico.

O gráfico mantém cinco posições mesmo quando menos de cinco municípios atendem ao filtro. As posições sem observação ficam vazias, evitando confundir ausência de municípios elegíveis com nota igual a zero.


```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(3, 2, figsize=(14, 12))
axes = axes.flatten()

ordem = ["REDACAO", "CN", "CH", "LC", "MT"]
cores = {"REDACAO": "steelblue", "CN": "darkorange",
         "CH": "seagreen", "LC": "indianred", "MT": "purple"}

for i, disciplina in enumerate(ordem):
    ax = axes[i]

    if disciplina not in top5_media or top5_media[disciplina].empty:
        ax.axis("off")
        ax.set_title(f"{disciplina} — sem dados", fontsize=11)
        continue

    df_plot = top5_media[disciplina].copy()
    coluna_media = [c for c in df_plot.columns if c.startswith("media_")][0]

    # Garante 5 linhas (se faltar, preenche com NaN -> barra vazia)
    if len(df_plot) < 5:
        faltam = 5 - len(df_plot)
        df_plot = pd.concat(
            [df_plot,
             pd.DataFrame({"NO_MUNICIPIO_TERRITORIO": ["—"] * faltam,
                           coluna_media: [float("nan")] * faltam})],
            ignore_index=True
        )

    df_plot = df_plot.sort_values(coluna_media, ascending=False, na_position="first")

    sns.barplot(
        data=df_plot,
        x=coluna_media,
        y="NO_MUNICIPIO_TERRITORIO",
        ax=ax,
        color=cores.get(disciplina, "steelblue")
    )

    for j, (_, row) in enumerate(df_plot.iterrows()):
        if pd.notna(row[coluna_media]):
            ax.text(row[coluna_media] + 0.3, j,
                    f"{row[coluna_media]:.1f}", va="center", fontsize=9)

    ax.set_title(f"Top 5 — {disciplina}", fontsize=12, fontweight="bold")
    ax.set_xlabel("Média")
    ax.set_ylabel("")

    vals = df_plot[coluna_media].dropna()
    if not vals.empty:
        ax.set_xlim(vals.min() * 0.97, vals.max() * 1.03)

axes[5].axis("off")

fig.suptitle(
    f"Top 5 municípios por média — {ANO_TRANSVERSAL}  "
    f"(N válidos > {N_VALIDOS_MIN})",
    fontsize=14, fontweight="bold", y=1.00
)

plt.tight_layout()
plt.show()
```

## 7. Outros indicadores para Top 5

Complementa os rankings com indicadores de participação e composição por dependência administrativa da escola.

A base aplica o filtro `n_participantes > 100` no nível municipal. Esse critério difere do filtro da seção 6, que considera `n_validos` por disciplina.

#### 7.1. Média simples das cinco disciplinas — somente 2009–2025

Calcula a média aritmética simples das médias municipais de Redação, CN, CH, LC e MT. O código exige que as cinco notas estejam disponíveis (`skipna=False`). Para 1998–2008, o indicador não é calculado porque a estrutura metodológica é diferente.

#### 7.2. Proporção de escola privada e pública

Compara os maiores percentuais municipais de participantes em escolas privadas e públicas. A proporção pública é a soma das proporções federal, estadual e municipal, exigindo que as três estejam disponíveis.

Os resultados descrevem a composição dos participantes por dependência administrativa, não a quantidade de escolas existentes em cada município.


```python
# ============================================================
# 7. OUTROS INDICADORES — TOP 5
# ============================================================
N_VALIDOS_MIN = 100  # mesmo limiar do bloco 6

colunas_indicadores = [
    "NO_MUNICIPIO_TERRITORIO", "n_participantes", *colunas_medias,
    "pct_escola_privada", "pct_escola_federal", "pct_escola_estadual",
    "pct_escola_municipal", "n_dependencia_informada",
]
indicadores_top5 = (
    dados_ano[[c for c in colunas_indicadores if c in dados_ano.columns]]
    .copy()
)

# ---- filtro comum: N participantes > limiar ----
if "n_participantes" not in indicadores_top5.columns:
    print(f"'n_participantes' ausente — não é possível aplicar filtro > {N_VALIDOS_MIN}.")
    indicadores_top5 = indicadores_top5.iloc[0:0]
else:
    n_antes = len(indicadores_top5)
    indicadores_top5 = (
        indicadores_top5[indicadores_top5["n_participantes"] > N_VALIDOS_MIN]
        .reset_index(drop=True)
    )
    print(f"Filtro n_participantes > {N_VALIDOS_MIN}: "
          f"{n_antes} → {len(indicadores_top5)} municípios.")

# ------------------------------------------------------------
# 7.1. MÉDIA SIMPLES DAS CINCO DISCIPLINAS (só 2009–2025)
# ------------------------------------------------------------
top5_media_5_areas = pd.DataFrame()

if PERIODO_ID_TRANSVERSAL == "2009_2025":
    colunas_areas = ["media_redacao", "media_cn", "media_ch", "media_lc", "media_mt"]
    if all(c in indicadores_top5.columns for c in colunas_areas) and not indicadores_top5.empty:
        indicadores_top5["media_5_areas"] = indicadores_top5[colunas_areas].mean(
            axis=1, skipna=False
        )
        top5_media_5_areas = (
            indicadores_top5[["NO_MUNICIPIO_TERRITORIO", "media_5_areas", "n_participantes"]]
            .dropna(subset=["media_5_areas"])
            .sort_values("media_5_areas", ascending=False)
            .head(5)
            .reset_index(drop=True)
        )
        print(f"\nTOP 5 — MÉDIA SIMPLES DAS CINCO DISCIPLINAS "
              f"(n_participantes > {N_VALIDOS_MIN})")
        display(top5_media_5_areas.round(2))
    else:
        print("Colunas de área ausentes ou base vazia após filtro — média das 5 áreas ignorada.")
else:
    print("Média simples das cinco áreas não calculada: 1998–2008 utiliza OBJETIVA.")

# ------------------------------------------------------------
# 7.2. PROPORÇÃO DE ESCOLA PRIVADA / PÚBLICA
# ------------------------------------------------------------
colunas_pct = ["pct_escola_federal", "pct_escola_estadual", "pct_escola_municipal"]

if all(c in indicadores_top5.columns for c in colunas_pct) and not indicadores_top5.empty:
    indicadores_top5["pct_escola_publica"] = indicadores_top5[colunas_pct].sum(
        axis=1, min_count=3
    )

    for coluna, rotulo in [
        ("pct_escola_privada",  "TOP 5 — PROPORÇÃO DE ESCOLA PRIVADA"),
        ("pct_escola_publica",  "TOP 5 — PROPORÇÃO DE ESCOLA PÚBLICA"),
    ]:
        if coluna not in indicadores_top5.columns:
            continue
        base = (
            indicadores_top5[["NO_MUNICIPIO_TERRITORIO", coluna, "n_participantes"]]
            .dropna(subset=[coluna])
        )
        if base.empty:
            print(f"\n{rotulo} — sem municípios válidos.")
            continue
        print(f"\n{rotulo} (n_participantes > {N_VALIDOS_MIN})")
        display(
            base.sort_values(coluna, ascending=False)
                .head(5)
                .reset_index(drop=True)
                .round(2)
        )
else:
    print("Colunas de proporção por dependência ausentes — seção 7.2 ignorada.")
```

## 10. Escola × desempenho

Investiga diferenças de desempenho segundo a dependência administrativa da escola, no nível município × grupo escolar × disciplina.

Considera as categorias `FEDERAL`, `ESTADUAL`, `MUNICIPAL`, `PRIVADA` e `NAO_INFORMADA`, conforme os registros disponíveis. O filtro `n_validos > 100` é aplicado a cada combinação antes da síntese estadual.

Se o produto de grupos não tiver dados para o ano selecionado, a seção informa a indisponibilidade, sem recorrer a outro ano.


```python
# ============================================================
# 10. ESCOLA × DESEMPENHO
# ============================================================
N_VALIDOS_MIN = 100  # mesmo limiar dos blocos 6 e 7

if not TEM_GRUPOS_ANO:

    escola_ano = pd.DataFrame()
    print(f"Sem dados de ESCOLA para {ANO_TRANSVERSAL} "
          "no produto desempenho_grupos_municipal.")

else:

    escola_ano = (
        grupo[
            (grupo["ano"] == ANO_TRANSVERSAL)
            & (grupo["tipo_grupo"].str.upper().eq("ESCOLA"))
            & (grupo["disciplina"].isin(ORDEM_DISCIPLINAS))
            & (grupo["n_validos"] > N_VALIDOS_MIN)          # ← filtro novo
        ]
        .copy()
        [["NO_MUNICIPIO_TERRITORIO",
          "grupo",
          "disciplina",
          "n_validos",
          "media",
          "mediana",
          "std"]]
        .sort_values(["disciplina", "grupo", "NO_MUNICIPIO_TERRITORIO"])
        .reset_index(drop=True)
    )

    n_total = (
        grupo[
            (grupo["ano"] == ANO_TRANSVERSAL)
            & (grupo["tipo_grupo"].str.upper().eq("ESCOLA"))
            & (grupo["disciplina"].isin(ORDEM_DISCIPLINAS))
        ].shape[0]
    )
    print(f"Filtro n_validos > {N_VALIDOS_MIN}: "
          f"{n_total} → {len(escola_ano)} linhas (escola × disciplina × município).")

    display(escola_ano)
```

## 10.1 Síntese estadual — Escola × desempenho

Agrega os resultados municipais por dependência administrativa e disciplina. A média é **ponderada pelo número de notas válidas (`n_validos`)**, evitando atribuir o mesmo peso a municípios com quantidades muito diferentes de observações.

A tabela também apresenta o total de notas válidas, a mediana das medianas municipais, a mediana dos desvios-padrão municipais e o número de municípios considerados. A função `media_ponderada_grupo()` é definida aqui e reutilizada nas análises por sexo e faixa etária.


```python
# ============================================================
# 10.1. SÍNTESE ESTADUAL — ESCOLA × DESEMPENHO
# ============================================================

def media_ponderada_grupo(df):

    peso = df["n_validos"]

    validos = (df["media"].notna()
               & peso.notna()
               & (peso > 0))

    if not validos.any():
        return np.nan

    return ((df.loc[validos, "media"] * peso.loc[validos]).sum()
            / peso.loc[validos].sum())


if escola_ano.empty:
    escola_resumo = pd.DataFrame()
    print("Sem linhas após filtro — síntese estadual não calculada.")
else:
    escola_resumo = (
        escola_ano
        .groupby(["grupo", "disciplina"], dropna=False)
        .apply(
            lambda g: pd.Series({
                "n_validos":          g["n_validos"].sum(),
                "media":              media_ponderada_grupo(g),
                "mediana_municipal":  g["mediana"].median(),
                "std_municipal":      g["std"].median(),
                "n_municipios":       g["NO_MUNICIPIO_TERRITORIO"].nunique(),
            }),
            include_groups=False,
        )
        .reset_index()
        .sort_values(["disciplina", "grupo"])
        .reset_index(drop=True)
    )
    display(escola_resumo.round(2))
```

Compara visualmente as médias ponderadas por dependência administrativa, em painéis separados por disciplina. A ordem das categorias é padronizada.

Os rótulos sobre as barras mostram a média; os rótulos do eixo horizontal informam quantos municípios contribuíram após o filtro `n_validos > 100`.

`Y_DESDE_ZERO` controla o início do eixo vertical. Quando `False`, o eixo é ajustado ao intervalo observado para destacar diferenças; essa escolha deve ser considerada na interpretação visual.

## 12. Faixa etária × desempenho

Analisa a composição etária dos participantes e as diferenças de desempenho entre as cinco faixas etárias analíticas do projeto.

A variável original `TP_FAIXA_ETARIA` é convertida para os grupos do projeto por `MAPA_FAIXA_5`. Os códigos originais representam categorias codificadas, não idades numéricas diretamente interpretáveis.

A seção separa a distribuição de participantes da análise de notas por faixa etária. Os filtros e níveis de agregação são descritos em cada subseção.


### 12.1 Proporção dos participantes por faixa etária

Soma `n_participantes` entre municípios para calcular a participação relativa de cada faixa. O filtro `n_participantes > N_VALIDOS_MIN` é aplicado no nível município × faixa antes da soma. O valor definido no código é `5`.

A tabela apresenta a contagem considerada e o percentual de cada faixa em relação ao total que permaneceu após o filtro. A soma dos percentuais é conferida ao final.


```python
# ============================================================
# 12.1A. DISTRIBUIÇÃO ETÁRIA DOS PARTICIPANTES
# ============================================================
N_VALIDOS_MIN = 5  

faixa_ano = faixa[faixa["ano"] == ANO_TRANSVERSAL].copy()

faixa_ano["grupo_faixa"] = (
    pd.to_numeric(faixa_ano["TP_FAIXA_ETARIA"], errors="coerce")
    .map(MAPA_FAIXA_5)
)

# ---- filtro N>100 NO NÍVEL DO MUNICÍPIO × FAIXA, antes de agregar ----
# (aqui n_participantes É a contagem do município naquela faixa)
faixa_ano = faixa_ano[
    faixa_ano["grupo_faixa"].notna()
    & (faixa_ano["n_participantes"] > N_VALIDOS_MIN)
].copy()

if faixa_ano.empty:
    print(f"Sem linhas com n_participantes > {N_VALIDOS_MIN} "
          f"para {ANO_TRANSVERSAL}.")
    proporcao_faixa = pd.DataFrame()
else:
    n_antes = len(faixa[faixa["ano"] == ANO_TRANSVERSAL])
    print(f"Filtro n_participantes > {N_VALIDOS_MIN} (nível município × faixa): "
          f"{n_antes} → {len(faixa_ano)} linhas.")

    proporcao_faixa = (
        faixa_ano.groupby("grupo_faixa", as_index=False)["n_participantes"]
        .sum()
    )

    proporcao_faixa["pct_participantes"] = (
        proporcao_faixa["n_participantes"]
        / proporcao_faixa["n_participantes"].sum() * 100
    )

    proporcao_faixa["grupo_faixa"] = pd.Categorical(
        proporcao_faixa["grupo_faixa"],
        categories=ORDEM_FAIXAS, ordered=True,
    )

    proporcao_faixa = (
        proporcao_faixa.sort_values("grupo_faixa").reset_index(drop=True)
    )

    print(f"\nDistribuição etária dos participantes — {ANO_TRANSVERSAL}")
    display(proporcao_faixa.round(2))

    print(f"\nTotal considerado: {proporcao_faixa['n_participantes'].sum():,.0f}")
    print(f"Soma dos percentuais: {proporcao_faixa['pct_participantes'].sum():.2f}%")
```

### 12.1B Desempenho por faixa etária × disciplina

Extrai as médias municipais por faixa etária e disciplina a partir de `desempenho_grupos_municipal`. Os grupos são harmonizados com `MAPA_FAIXA_5`.

O filtro `n_validos > N_VALIDOS_MIN` é aplicado no nível município × faixa × disciplina, antes da agregação. Para cada combinação, são somados os participantes válidos e calculada a média ponderada.

Se não houver dados para o ano ou nenhuma linha permanecer após o filtro, o resultado é mantido vazio e a indisponibilidade é informada.

```python
# ============================================================
# 12.1B. DESEMPENHO POR FAIXA ETÁRIA × DISCIPLINA
# ============================================================
if not TEM_GRUPOS_ANO:

    faixa_desempenho = pd.DataFrame()
    print(f"Sem dados de FAIXA_ETARIA para {ANO_TRANSVERSAL} "
          "no produto desempenho_grupos_municipal.")

else:

    base_bruta = grupo[
        (grupo["ano"] == ANO_TRANSVERSAL)
        & (grupo["tipo_grupo"].str.upper().eq("FAIXA_ETARIA"))
        & (grupo["disciplina"].isin(ORDEM_DISCIPLINAS))
    ].copy()

    base_bruta["grupo_faixa"] = (
        pd.to_numeric(base_bruta["grupo"], errors="coerce")
        .map(MAPA_FAIXA_5)
    )

    base_bruta = base_bruta[base_bruta["grupo_faixa"].notna()]

    # ---- filtro N>100 LINHA A LINHA, antes do groupby ----
    n_antes = len(base_bruta)
    base_filtrada = base_bruta[base_bruta["n_validos"] > N_VALIDOS_MIN].copy()
    print(f"Filtro n_validos > {N_VALIDOS_MIN} (município × faixa × disciplina): "
          f"{n_antes} → {len(base_filtrada)} linhas.")

    if base_filtrada.empty:
        faixa_desempenho = pd.DataFrame()
        print("Sem linhas após filtro — desempenho por faixa não calculado.")
    else:
        faixa_desempenho = (
            base_filtrada
            .groupby(
                ["NO_MUNICIPIO_TERRITORIO", "grupo_faixa", "disciplina"],
                dropna=False,
            )
            .apply(
                lambda g: pd.Series({
                    "n_validos": g["n_validos"].sum(),
                    "media":     media_ponderada_grupo(g),
                }),
                include_groups=False,
            )
            .reset_index()
            .sort_values(["disciplina", "grupo_faixa", "NO_MUNICIPIO_TERRITORIO"])
            .reset_index(drop=True)
        )
        display(faixa_desempenho.round(2))
```

Consolida os resultados municipais por faixa etária e disciplina. A média é ponderada pelo número de notas válidas; `mediana_municipal` resume a distribuição das médias municipais.

A tabela também informa o total de notas válidas e o número de municípios representados. A ordenação respeita `ORDEM_FAIXAS`.


```python
# ============================================================
# 12.1. SÍNTESE ESTADUAL — FAIXA ETÁRIA × DESEMPENHO
# ============================================================

if faixa_desempenho.empty:

    faixa_resumo = pd.DataFrame()

    print(f"Síntese indisponível para {ANO_TRANSVERSAL}.")

else:

    faixa_resumo = (
        faixa_desempenho
        .groupby(
            ["grupo_faixa", "disciplina"],
            dropna=False)
        .apply(
            lambda g: pd.Series({
                "n_validos": g["n_validos"].sum(),
                "media": media_ponderada_grupo(g),
                "mediana_municipal": g["media"].median(),
                "n_municipios":
                    g["NO_MUNICIPIO_TERRITORIO"].nunique(),
            }),
            include_groups=False
        ).reset_index())

    faixa_resumo["grupo_faixa"] = pd.Categorical(faixa_resumo["grupo_faixa"],categories=ORDEM_FAIXAS,ordered=True)
    faixa_resumo = (faixa_resumo.sort_values(["disciplina", "grupo_faixa"]).reset_index(drop=True))

    display(faixa_resumo.round(2))
```

```python
# ============================================================
# 12.2. PERFIL ETÁRIO DO DESEMPENHO
# ============================================================
sns.set_theme(style="whitegrid")

cores_disc = {
    "REDACAO": "purple",
    "CN":      "green",
    "CH":      "tomato",
    "LC":      "grey",
    "MT":      "dodgerblue",
}

if faixa_resumo.empty:
    print(f"Visualização indisponível para {ANO_TRANSVERSAL}: "
          "não há dados de desempenho por faixa etária.")
else:
    ordem_disciplinas = [d for d in ORDEM_DISCIPLINAS
                         if d in faixa_resumo["disciplina"].unique()]
    ordem_faixas = [f for f in ORDEM_FAIXAS
                    if f in faixa_resumo["grupo_faixa"].astype(str).unique()]

    if not ordem_disciplinas or not ordem_faixas:
        print("Nenhuma disciplina ou faixa disponível em faixa_resumo.")
    else:
        fig, ax = plt.subplots(figsize=(13, 6.5))

        sns.barplot(
            data=faixa_resumo,
            x="grupo_faixa",
            y="media",
            hue="disciplina",
            order=ordem_faixas,
            hue_order=ordem_disciplinas,
            palette=cores_disc,
            dodge=True,                    # barras lado a lado (padrão p/ hue)
            errorbar=None,                 # médias já ponderadas em 12.1
            ax=ax,
        )

        # ---- rótulos nas barras ----
        # Em gráfico agrupado, ax.patches vem na ordem hue × x.
        # Só rotulamos se houver poucas barras, para não poluir.
        n_barras = len(ax.patches)
        rotular = n_barras <= 25

        if rotular:
            vals_globais = faixa_resumo["media"].dropna()
            span = (vals_globais.max() - vals_globais.min()) if len(vals_globais) > 1 else 1.0
            offset = max(span * 0.015, 0.5)

            for barra in ax.patches:
                valor = barra.get_height()
                if pd.isna(valor):
                    continue
                ax.text(
                    barra.get_x() + barra.get_width() / 2,
                    valor + offset,
                    f"{valor:.0f}",
                    ha="center", va="bottom", fontsize=7,
                )

        ax.set_xlabel("Faixa etária")
        ax.set_ylabel("Média da nota")
        ax.set_title(
            f"Desempenho médio por faixa etária — "
            f"{ANO_TRANSVERSAL} ({PERIODO_TRANSVERSAL['rotulo']})",
            fontsize=13, fontweight="bold",
        )
        ax.legend(title="Disciplina", bbox_to_anchor=(1.02, 1), loc="upper left")
        ax.grid(axis="y", alpha=.3)
        ax.tick_params(axis="x", rotation=0)

        # ---- ylim dinâmico: espaço p/ rótulo em cima ----
        vals = faixa_resumo["media"].dropna()
        if not vals.empty:
            ymin, ymax = float(vals.min()), float(vals.max())
            span_y = ymax - ymin
            if span_y <= 0:
                span_y = max(abs(ymax) * 0.1, 1.0)
            ax.set_ylim(ymin - span_y * 0.08, ymax + span_y * 0.15)

        plt.tight_layout()
        plt.show()
```

### 12.4 Distribuição etária dos participantes

Apresenta novamente a composição etária diretamente a partir de `faixa_etaria_municipal`, **sem aplicar o filtro mínimo** utilizado na subseção 12.1.

Por isso, os totais e percentuais podem diferir dos apresentados em 12.1, que exclui combinações município × faixa com até cinco participantes (conforme o parâmetro do código). A comparação entre as duas saídas permite observar o efeito desse filtro.


```python
# ============================================================
# 12.3. DISTRIBUIÇÃO ETÁRIA DOS PARTICIPANTES
# ============================================================

faixa_ano = faixa[faixa["ano"] == ANO_TRANSVERSAL].copy()

if faixa_ano.empty:

    print(f"Sem dados de faixa etária para {ANO_TRANSVERSAL}.")

else:

    faixa_ano["grupo_faixa"] = (
        pd.to_numeric(
            faixa_ano["TP_FAIXA_ETARIA"],
            errors="coerce"
        )
        .map(MAPA_FAIXA_5)
    )

    perfil_etario_participacao = (
        faixa_ano[
            faixa_ano["grupo_faixa"].notna()
        ]
        .groupby(
            "grupo_faixa"
        )["n_participantes"]
        .sum()
        .reindex(ORDEM_FAIXAS)
        .rename("n_participantes")
        .reset_index()
    )

    perfil_etario_participacao["pct"] = (
        perfil_etario_participacao["n_participantes"]
        / perfil_etario_participacao["n_participantes"].sum()* 100
    )

    display(perfil_etario_participacao.round(2))
```

### 13.1 Participação × desempenho

Produz um gráfico de dispersão entre o número de participantes e a média municipal de Redação. O tamanho dos pontos também representa o número de participantes; o eixo horizontal usa escala logarítmica para acomodar diferenças de magnitude.

O código identifica uma disciplina complementar disponível (a primeira disciplina não correspondente à Redação), mas o eixo vertical permanece associado à média de Redação. A variável complementar é informada no texto de saída, não representada como segunda dimensão do gráfico.

A visualização permite observar padrões e possíveis municípios discrepantes, mas não estabelece causalidade.


```python
# ============================================================
# 13.1. PARTICIPAÇÃO × DESEMPENHO
# ============================================================
# Usa a Redação do período e a primeira disciplina não-Redação disponível.
coluna_redacao = "media_redacao"
colunas_alternativas = [c for c in colunas_medias if c != coluna_redacao]
coluna_comparacao = colunas_alternativas[0] if colunas_alternativas else None

if coluna_redacao not in dados_ano.columns or coluna_comparacao is None:
    print("Gráfico participação × desempenho indisponível para esta estrutura.")
else:
    dados_participacao = dados_ano[
        ["NO_MUNICIPIO_TERRITORIO", "n_participantes", coluna_redacao, coluna_comparacao]
    ].dropna().copy()
    fig, ax = plt.subplots(figsize=(11, 7))
    sns.scatterplot(data=dados_participacao, x="n_participantes", y=coluna_redacao,
                    size="n_participantes", sizes=(40, 500), alpha=.7, ax=ax)
    ax.set_xscale("log")
    ax.set_xlabel("Número de participantes (escala log)")
    ax.set_ylabel(f"Média municipal — {nomes_disciplinas[coluna_redacao]}")
    ax.set_title(f"Participação e desempenho municipal — {ANO_TRANSVERSAL} ({PERIODO_TRANSVERSAL['rotulo']})")
    ax.grid(axis="both", alpha=.3)
    plt.tight_layout()
    plt.show()
    print(f"Variável complementar disponível: {nomes_disciplinas[coluna_comparacao]}")
```
