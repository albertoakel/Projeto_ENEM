### inspecionar os CSVs de 2024

```python

from pathlib import Path
import csv

from enem_pipeline.config import STAGING_DIR

diretorio = STAGING_DIR / "ano=2024"

for nome in ["PARTICIPANTES_2024.csv", "RESULTADOS_2024.csv"]:
    caminho = diretorio / nome

    print(f"\n{'=' * 60}")
    print(nome)

    if not caminho.is_file():
        print("Arquivo não encontrado:", caminho)
        continue

    with caminho.open(
        "r",
        encoding="utf-8-sig",
        errors="replace",
        newline="",
    ) as arquivo:
        amostra = arquivo.read(65536)

    primeira_linha = amostra.splitlines()[0] if amostra else ""

    try:
        dialect = csv.Sniffer().sniff(
            amostra,
            delimiters=";,\t|",
        )
        separador = dialect.delimiter
    except csv.Error:
        separador = ";"

    cabecalho = next(
        csv.reader([primeira_linha], delimiter=separador),
        [],
    )

    print("Separador:", repr(separador))
    print("Quantidade de colunas:", len(cabecalho))
    print("Primeiras 20 colunas:", cabecalho[:20])
    print("Coluna SG_UF_PROVA:", "SG_UF_PROVA" in cabecalho)
    print("Coluna NU_INSCRICAO:", "NU_INSCRICAO" in cabecalho)
```
