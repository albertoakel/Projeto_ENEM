# 12 — Camada de Consumo

Constrói o painel municipal e os produtos auxiliares a partir dos Parquets Analytic. Não realiza junções individuais.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display

PROJECT_ROOT = Path.cwd().resolve()
while not (PROJECT_ROOT / "src" / "enem_pipeline").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
if not (PROJECT_ROOT / "src" / "enem_pipeline").exists():
    raise FileNotFoundError("Não encontrei src/enem_pipeline. Abra o notebook a partir do projeto ENEM.")
sys.path.insert(0, str(PROJECT_ROOT / "src"))

from enem_pipeline import config as cfg
from enem_pipeline.consumo import executar_camada_consumo

UF_PROCESSAMENTO = "RR"
METADATA_DIR = Path(cfg.METADATA_DIR)
ANALYTIC_MANIFESTO = METADATA_DIR / f"manifesto_analytic_{UF_PROCESSAMENTO}.csv"
CONSUMO_ROOT = Path(cfg.CONSUMO_DIR)

print("Projeto:", PROJECT_ROOT)
print("UF:", UF_PROCESSAMENTO)
print("Manifesto Analytic:", ANALYTIC_MANIFESTO)
print("Destino Consumo:", CONSUMO_ROOT / f"uf={UF_PROCESSAMENTO}")
```

## Execução completa e auditoria

```python
resultado_consumo = executar_camada_consumo(
    manifesto_analytic_path=ANALYTIC_MANIFESTO,
    metadata_dir=METADATA_DIR,
    consumo_root=CONSUMO_ROOT,
    uf=UF_PROCESSAMENTO,
    gravar=True,
)

display(resultado_consumo["catalogo"])
display(resultado_consumo["auditoria"])
print(f"Tempo: {resultado_consumo['duracao_segundos']:.2f} s")
print("Diretório:", resultado_consumo["diretorio_consumo"])
print("AUDITORIA:", "OK" if resultado_consumo["auditoria_ok"] else "COM FALHAS")
if not resultado_consumo["auditoria_ok"]:
    raise AssertionError("Auditoria da camada de Consumo falhou; consulte a tabela acima.")
```
