# 10 — Camada Analítica

Este notebook executa a rotina em `src/enem_pipeline/analytic.py`. Não modifica Gold. Os produtos são agregados por município e ano; não há junção individual.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display

# Procura a raiz do projeto a partir do diretório atual.
PROJECT_ROOT = Path.cwd().resolve()
while not (PROJECT_ROOT / "src" / "enem_pipeline").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
if not (PROJECT_ROOT / "src" / "enem_pipeline").exists():
    raise FileNotFoundError("Não encontrei src/enem_pipeline. Abra o notebook a partir do projeto ENEM.")
sys.path.insert(0, str(PROJECT_ROOT / "src"))

from enem_pipeline import config as cfg
from enem_pipeline.analytic import executar_camada_analitica

# UF que será processada nesta execução.
UF_PROCESSAMENTO = "RR"
ANO_INICIAL = int(getattr(cfg, "ANO_INICIAL", 1998))
ANO_FINAL = int(getattr(cfg, "ANO_FINAL", 2025))
METADATA_DIR = Path(cfg.METADATA_DIR)
GOLD_MANIFESTO = METADATA_DIR / f"manifesto_gold_{UF_PROCESSAMENTO}.csv"
ANALYTIC_ROOT = Path(cfg.ANALYTIC_DIR)

print("Projeto:", PROJECT_ROOT)
print("UF:", UF_PROCESSAMENTO)
print("Manifesto Gold:", GOLD_MANIFESTO)
print("Destino Analytic:", ANALYTIC_ROOT / f"uf={UF_PROCESSAMENTO}")
```

## Execução completa

```python
resultado_analytic = executar_camada_analitica(
    manifesto_gold_path=GOLD_MANIFESTO,
    metadata_dir=METADATA_DIR,
    analytic_root=ANALYTIC_ROOT,
    uf=UF_PROCESSAMENTO,
    ano_inicial=ANO_INICIAL,
    ano_final=ANO_FINAL,
    gravar=True,
)

print(f"Fontes Gold auditadas: {len(resultado_analytic['fontes_auditadas'])}")
print(f"Produtos analíticos: {len(resultado_analytic['produtos'])}")
print(f"Tempo: {resultado_analytic['duracao_segundos']:.2f} s")
print("Diretório:", resultado_analytic["diretorio_analytic"])
display(resultado_analytic["fontes_auditadas"])
display(resultado_analytic["catalogo"])
display(resultado_analytic["manifesto"])
```

## Verificação rápida dos produtos

```python
resumo = []
for nome, df in resultado_analytic["produtos"].items():
    resumo.append({
        "produto": nome,
        "linhas": len(df),
        "colunas": len(df.columns),
        "anos": f"{df['ano'].min()}–{df['ano'].max()}" if not df.empty else "vazio",
        "municipios": df["CO_MUNICIPIO_TERRITORIO"].nunique() if "CO_MUNICIPIO_TERRITORIO" in df else 0,
    })
display(pd.DataFrame(resumo))
```

## Observações de interpretação
- `n_participantes` representa a contagem de linhas da fonte Gold, sem deduplicação automática.
- Estatísticas de notas consideram valores não nulos; a quantidade válida é explicitada.
- 2024–2025 usam `participantes` para demografia/faixa etária e `resultados` para presença, notas, escola e grupos.
- O produto de grupos respeita a disponibilidade histórica de sexo, escola, faixa etária e disciplinas.
