## **Distribuição e desigualdade do desempenho**

**Área de análise:** Pará (PA)\
**Referência territorial:** município da escola (ESC)\
**Período:** 1998--2025 para o painel municipal e os grupos escolares;
2001--2023 para o detalhamento por sexo e faixa etária.

O Eixo 6 investiga como as notas do ENEM se distribuem ao longo do
tempo, se a dispersão dos resultados aumentou ou diminuiu, qual é a
distância entre os segmentos de menor e maior desempenho e como essas
diferenças se manifestam entre municípios, disciplinas e grupos de
participantes.

A análise utiliza quatro fontes: o painel municipal de desempenho
(`painel`), os resultados agregados por grupos (`grupo`), o detalhamento
por sexo e faixa etária (`dsp`) e a composição dos participantes por
idade (`faixa`). A fonte `faixa` informa a composição etária, mas não
contém notas; portanto, não é utilizada para calcular desempenho.

**Ressalva metodológica central:** as médias apresentadas são ponderadas
pelo número de notas válidas. Já medianas, quartis e IQR apresentados no
painel são resumos das estatísticas municipais. Não representam os
quartis ou a mediana calculados diretamente sobre todas as notas
individuais dos participantes do Pará. O desvio-padrão combinado é uma
aproximação construída a partir das estatísticas municipais e deve ser
interpretado com essa ressalva.

Também é importante esclarecer a escala: os resultados anteriores a 2009
já aparecem na planilha em magnitudes compatíveis com a escala 0--1000.
Não se deve multiplicar novamente esses valores por 10 A estrutura
do ENEM mudou em 2009; por isso, a prova Objetiva anterior a 2009 é
apresentada separadamente das áreas do Novo ENEM

## 1. Como evoluiu a dispersão das notas ao longo dos anos?

```python
serie_anual = eda2.resumir_distribuicao_municipal_eixo6(desempenho_municipal)
ANO_INICIO = 2002      # 1998-2001 tem poucos municípios (5 a 116)

def plot_nivel(g, disciplina):
    g = g[g.ano >= ANO_INICIO].sort_values("ano")
    x = g["ano"]

    fig, ax = plt.subplots(figsize=(12, 4.2))

    ax.fill_between(x, g.p25_medias_municipais, g.p75_medias_municipais,
                    color="#9ecae1", alpha=0.35, label="P25-P75 das médias municipais")
    ax.plot(x, g.media_ponderada_sem_zero, marker="o", ms=3.5, lw=1.8,
            color="#08306b", label="Média ponderada (sem zeros)")
    ax.plot(x, g.media_ponderada, ls="--", lw=1.2, color="#6baed6",
            label="Média ponderada (com zeros)")
    ax.plot(x, g.mediana_das_medianas_municipais, ls=":", lw=1.4,
            color="#d62728", label="Mediana das medianas municipais")

    ax.axvspan(2019.5, 2021.5, color="gray", alpha=0.12)          # pandemia
    if x.min() < 2008.5 < x.max():
        ax.axvline(2008.5, color="gray", ls="--", alpha=0.7)       # mudança de escala/prova
    ax.set_xticks(range(int(x.min()), int(x.max()) + 1, 2))
    ax.set_xlabel("Ano", fontsize=8)
    ax.set_ylabel("Nota (0-1000)", fontsize=8)
    ax.set_title(f"{disciplina}: nível e faixa central das médias municipais "
                 f"(faixa cinza = 2020-2021)", fontsize=10, fontweight="bold", loc="left")
    ax.tick_params(labelsize=7)
    ax.grid(alpha=0.3)
    ax.legend(frameon=False, fontsize=7, ncol=2, loc="upper left")
    plt.tight_layout()
    plt.show()

for disciplina, g in serie_anual.groupby("disciplina"):
    plot_nivel(g, disciplina)

```

## 5. Detalhamento complementar esempenho e composição etária

```python
dsp_resumo = dsp_preparado.copy()
dsp_resumo["_soma"] = dsp_resumo["media"] * dsp_resumo["n_validos"]
dsp_resumo["_n_sem_zero"] = (dsp_resumo["n_validos"] - dsp_resumo["n_zeros"].fillna(0)).clip(lower=0)
dsp_resumo["_soma_sem_zero"] = dsp_resumo["media_positivas"] * dsp_resumo["_n_sem_zero"]
dsp_resumo = dsp_resumo.groupby(["ano", "disciplina", "sexo", "faixa_etaria"], as_index=False).agg(
    n_municipios=("CO_MUNICIPIO_TERRITORIO", "nunique"),
    n_validos=("n_validos", "sum"), n_zeros=("n_zeros", "sum"),
    soma_media=("_soma", "sum"), n_sem_zero=("_n_sem_zero", "sum"),
    soma_media_sem_zero=("_soma_sem_zero", "sum"),
)
dsp_resumo["media_ponderada"] = dsp_resumo["soma_media"] / dsp_resumo["n_validos"].where(dsp_resumo["n_validos"].gt(0))
dsp_resumo["media_ponderada_sem_zero"] = dsp_resumo["soma_media_sem_zero"] / dsp_resumo["n_sem_zero"].where(dsp_resumo["n_sem_zero"].gt(0))
#display(dsp_resumo)
#display(composicao_etaria.head(40))

ORDEM_FAIXA = ["<17", "17–20", "21–25", "26–40", ">40"]
DISC5 = ["CH", "CN", "LC", "MT", "REDACAO"]
CORES = {"CH": "#1f77b4", "CN": "#2ca02c", "LC": "#ff7f0e", "MT": "#9467bd", "REDACAO": "#d62728"}
ANO_PIRAMIDE = 2023
MIN_N = 100

d = dsp_resumo[dsp_resumo.ano >= 2009].copy()
d["faixa_etaria"] = pd.Categorical(d.faixa_etaria, categories=ORDEM_FAIXA, ordered=True)

w = (d[d.n_sem_zero >= MIN_N]
     .pivot_table(index=["ano", "sexo", "faixa_etaria"], columns="disciplina",
                  values="media_ponderada_sem_zero", observed=True)[DISC5].dropna().reset_index())

g = sns.pairplot(w, vars=DISC5, hue="faixa_etaria", hue_order=ORDEM_FAIXA,
                 palette="viridis", corner=True, height=2.0,
                 plot_kws={"s": 18, "alpha": 0.75, "edgecolor": "none"},
                 diag_kws={"common_norm": False})
g.figure.suptitle("Média sem zeros por grupo (sexo × faixa etária × ano): disciplina contra disciplina",
                  y=1.02, fontsize=12, fontweight="bold")
for ax in g.axes.flat:
    if ax is not None:
        ax.tick_params(labelsize=7); ax.xaxis.label.set_size(8); ax.yaxis.label.set_size(8)
plt.show()

#print(w[DISC5].corr().round(2))     # correlação entre disciplinas
ANO_INI, ANO_FIM = 2009, 2023

d = dsp_resumo[(dsp_resumo.ano >= ANO_INI) & (dsp_resumo.ano <= ANO_FIM)]
g = (d.groupby(["faixa_etaria", "disciplina"])
       .agg(soma=("soma_media_sem_zero", "sum"), n=("n_sem_zero", "sum")).reset_index())
g["media"] = g.soma / g.n

media = g.pivot(index="faixa_etaria", columns="disciplina", values="media").loc[ORDEM_FAIXA, DISC5]
n = g.pivot(index="faixa_etaria", columns="disciplina", values="n").loc[ORDEM_FAIXA, DISC5]
rotulo = media.round(0).astype(int).astype(str) + "\n(n=" + (n / 1000).round(1).astype(str) + " mil)"

# Cor = média da faixa menos a média da disciplina (compara faixas dentro da prova)
dif = media - media.mean()

fig, ax = plt.subplots(figsize=(12, 6))
sns.heatmap(dif, annot=rotulo, fmt="", cmap="RdBu", center=0, linewidths=.5,
            linecolor="white", annot_kws={"fontsize": 8},
            cbar_kws={"label": f'Diferença para a média das \n faixas, na disciplina (pts)'}, ax=ax)
ax.set_title(f"Desempenho por faixa etária e disciplina ({ANO_INI}-{ANO_FIM})",
             fontsize=11, fontweight="bold")
ax.set_xlabel(""); ax.set_ylabel("")
ax.tick_params(labelsize=8); ax.figure.axes[-1].tick_params(labelsize=8)
plt.tight_layout(); plt.show()
```

Nas provas objetivas, as quedas são menores — por exemplo, em Matemática, vão de −23,0 pontos (21–25 anos) a −36,6 pontos (acima de 40 anos), e em Ciências Humanas, situam-se em −24,6 pontos e −19,3 pontos, respectivamente.

Em síntese, as diferenças etárias seguem uma tendência comum às cinco provas, mas com maior amplitude na Redação. Como os resultados são expressos em relação à média de cada disciplina e cada prova possui estrutura avaliativa e distribuição próprias, esses valores não devem ser tomados isoladamente como medida direta de dificuldade nem como evidência causal entre idade e desempenho.


## 6-Síntese dos resultados


A análise da distribuição do desempenho no ENEM no Pará evidencia que as diferenças observadas não seguem uma tendência única. A variabilidade das notas, a distribuição das médias municipais e as diferenças entre grupos de participantes apresentam comportamentos distintos conforme a disciplina e o período considerado. Esse conjunto de resultados reforça a importância de analisar o desempenho para além das médias gerais, incorporando também medidas de dispersão e características da composição dos grupos avaliados.

Entre 2009 e 2025, a evolução dos indicadores de dispersão foi distinta entre as áreas do conhecimento. Ciências Humanas, Ciências da Natureza e Linguagens apresentaram redução do desvio-padrão combinado aproximado, embora os indicadores baseados no IQR municipal não tenham acompanhado essa redução de maneira uniforme. Matemática apresentou comportamento diferente, com ampliação expressiva da dispersão em relação a 2009. Apesar do recuo observado entre 2023 e 2025, seus indicadores permaneceram acima dos valores iniciais da série. Na Redação, houve redução da dispersão em relação a 2009, mas os resultados ainda apresentam variabilidade elevada quando comparados às provas objetivas.

As mudanças na distribuição das médias municipais também não foram homogêneas. Em Ciências Humanas e Ciências da Natureza, os quartis inferior e superior aumentaram moderadamente entre 2009 e 2025. Em Linguagens, a elevação foi mais expressiva, enquanto a Redação apresentou os maiores avanços nos dois quartis, acompanhados de redução da distância entre eles. Matemática, por sua vez, registrou diminuição do quartil inferior e pequena redução do superior, ampliando a distância entre essas referências. Portanto, a evolução das médias municipais indica trajetórias distintas entre as disciplinas, sem evidenciar uma convergência geral dos resultados territoriais.

As diferenças entre grupos de participantes acrescentam outra dimensão a esse quadro. Em 2023, as amplitudes medianas municipais entre faixas etárias foram superiores às observadas entre categorias de sexo e dependência administrativa nas cinco disciplinas, especialmente em Matemática e Redação. O detalhamento das médias por faixa etária também revelou um padrão consistente: os grupos abaixo de 17 anos e de 17–20 anos apresentaram desvios positivos em relação à média de cada disciplina, enquanto os grupos a partir de 21 anos registraram desvios negativos. Essa diferença foi particularmente acentuada na Redação. Embora esse comportamento seja observado nas cinco provas, os resultados não permitem atribuir à idade um efeito isolado sobre o desempenho, uma vez que as categorias também podem refletir diferenças nas trajetórias escolares e nos perfis de participação.

As comparações entre dependências administrativas mostram, adicionalmente, diferenças expressivas entre os grupos de escolas. Em 2025, as médias ponderadas das escolas privadas foram superiores às das escolas estaduais em todas as disciplinas, com maiores distâncias em Redação e Matemática. As escolas federais também apresentaram médias superiores às estaduais. Essas diferenças, entretanto, devem ser interpretadas como comparações entre grupos institucionais, considerando as limitações de cobertura de algumas categorias, particularmente das escolas municipais no último ano da série.

A comparação entre disciplinas evidencia que as diferenças de desempenho também precisam ser interpretadas à luz das características de cada prova. Em 2023, as médias ponderadas das provas objetivas foram relativamente próximas, variando de 479,0 pontos em Ciências da Natureza a 504,8 em Ciências Humanas. Matemática apresentou média de 504,5 pontos, enquanto Linguagens registrou 497,5. A Redação apresentou média superior, de 628,0 pontos. Entretanto, a proporção de notas zero foi consideravelmente maior nessa disciplina, alcançando aproximadamente 5,7% dos registros, contra menos de 0,6% nas provas objetivas. Quando consideradas apenas as notas positivas, a média ponderada da Redação passou para aproximadamente 665,9 pontos, enquanto as alterações nas demais disciplinas foram menores. Assim, a ocorrência de notas zero constitui um componente importante na interpretação da média agregada da Redação.

Em conjunto, os resultados mostram que a desigualdade de desempenho no ENEM no Pará se manifesta de diferentes formas. Matemática se destaca pelo aumento da dispersão em relação a 2009, enquanto a Redação apresenta maior amplitude das diferenças entre faixas etárias e maior influência das notas zero sobre a média agregada. Ciências Humanas, Ciências da Natureza e Linguagens apresentam redução de alguns indicadores de dispersão, mas esse comportamento não implica necessariamente uma diminuição de todas as diferenças territoriais ou entre grupos.
