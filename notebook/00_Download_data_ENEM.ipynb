## 1. library

```python
#configurações
from _setup import *
from enem_pipeline.config import PROJECT_ROOT,BRONZE_DIR
import subprocess

zip_dir = BRONZE_DIR
zip_dir.mkdir(parents=True, exist_ok=True)

ano_inicial = 1998
ano_final   = 2001
```

```python
# Execução & download

for ano in range(ano_inicial, ano_final):
    url =f'https://download.inep.gov.br/microdados/microdados_enem_{ano}.zip'
    filename = zip_dir / f'microdados_enem_{ano}.zip'

    if not filename.exists():
        print(f'Baixando ENEM {ano}...')
        cmd = f'curl -k -L -o {filename} {url}'
        result = subprocess.run(cmd, shell=True, capture_output=True)
        if result.returncode == 0:
            print(f'✓ ENEM {ano} baixado')
        else:
            print(f'✗ ENEM {ano} falhou (código {result.returncode})')
    else:
        print(f'ENEM {ano} já existe')
```
